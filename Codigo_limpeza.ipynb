```python
geolocation[geolocation.duplicated(keep=False)] #verificando se funcionou#
```

```python
geolocation.to_csv('geolocation_no_duplicates.csv', index=False) #salvando nova base#
print('Geolocation data without duplicates saved to geolocation_no_duplicates.csv')
```

```python
products = products.dropna(subset=['product_category_name']) #tirando os valores nulos de produto#
print('Shape of products DataFrame after removing rows with null product_category_name:')
display(products.shape)
```
